## $^{16}$ O Separation of the Pointcloud for target-like $\alpha$

```python
import h5py
import numpy as np
import matplotlib.pyplot as plt
%matplotlib widget
```

## $^{16}\text{O}$ Inelastic

```python
run_num = 0
simul_pc_cloud = f"/Volumes/research_EXT_2/engine_decay/Pointcloud/run_000{run_num}.h5"

simul_target_pc = h5py.File(f"/Volumes/research_EXT_2/spyral_target_alpha_width3/Pointcloud/run_000{run_num}.h5", "w")
simul_decay_pc = h5py.File(f"/Volumes/research_EXT_2/spyral_decay_alpha_width3/Pointcloud/run_000{run_num}.h5", "w")

file = h5py.File(simul_pc_cloud, "r")
ls = list(file.keys())[0]  # noqa: RUF015
data = file[ls]
min_event = data.attrs["min_event"]
max_event = data.attrs["max_event"]

target_group = simul_target_pc.create_group("cloud")
decay_group = simul_decay_pc.create_group("cloud")

target_group.attrs["min_event"] = min_event
target_group.attrs["max_event"] = max_event

decay_group.attrs["min_event"] = min_event
decay_group.attrs["max_event"] = max_event

counter = 0
five_tracks = 0


for key_num in range(min_event, max_event+1):
    
    key = f"cloud_{key_num}"
    if key not in data:
        continue
    
    label_name = f"labels_{key_num}"
    labels = data[label_name][:]
    
    data_all = data[key][:]
    
    labels = labels.squeeze()
    mask_target = labels == 2
    mask_decay = labels != 2
    counter+=1
    if len(np.unique(labels)) == 5:
        five_tracks+=1
    
    points_target = data_all[mask_target]
    points_decay = data_all[mask_decay]
    
    # fig = plt.figure(figsize=(6, 6), layout='constrained')
    # ax = fig.add_subplot(111, projection='3d')
    # ax.scatter(points_target[:, 2], points_target[:, 0], points_target[:, 1], color="crimson",s=3, label="Label 2")
    # ax.scatter(points_decay[:, 2], points_decay[:, 0], points_decay[:, 1], color="navy", s=3, label="Other Labels")
    # ax.grid(True)
    # ax.set_xlim3d(0, 1000)
    # ax.set_xlabel("Z [mm]")
    # ax.set_ylim3d(-300, 300)
    # ax.set_ylabel("X [mm]")

    # ax.set_zlim3d(-300, 300)
    # ax.set_zlabel("Y [mm]")
    
    # plt.legend()
    # plt.show()
    
    
    target_ds = target_group.create_dataset(
        key,
        data = points_target
    )
    
    target_group.create_dataset(
        label_name,
        data = labels[mask_target]
    )

    target_ds.attrs["orig_run"] =  run_num 
    target_ds.attrs["orig_event"] = int(key.strip("cloud_")) 
    
    target_ds.attrs["ic_amplitude"] = float(-1)
    target_ds.attrs["ic_integral"] = float(-1)
    target_ds.attrs["ic_centroid"] = float(-1)
    target_ds.attrs["ic_multiplicity"] = float(-1)
    target_ds.attrs["ic_sca_centroid"] = float(-1)
    target_ds.attrs["ic_sca_multiplicity"] = float(-1)
    
    decay_ds = decay_group.create_dataset(
        key,
        data = points_decay
    )
    
    decay_group.create_dataset(
        label_name,
        data = labels[mask_decay]
    )
    
    
    decay_ds.attrs["orig_run"] =  run_num 
    decay_ds.attrs["orig_event"] = int(key.strip("cloud_")) 
    
    decay_ds.attrs["ic_amplitude"] = float(-1)
    decay_ds.attrs["ic_integral"] = float(-1)
    decay_ds.attrs["ic_centroid"] = float(-1)
    decay_ds.attrs["ic_multiplicity"] = float(-1)
    decay_ds.attrs["ic_sca_centroid"] = float(-1)
    decay_ds.attrs["ic_sca_multiplicity"] = float(-1)
    
    
file.close()
simul_decay_pc.close()
simul_target_pc.close()
print(f"Number of inelasic (five) tracks (run {run_num}): {five_tracks} out of total {counter}")
```
