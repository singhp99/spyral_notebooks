```python
import pandas as pd
import h5py
import numpy as np
from pathlib import Path
```

## Combine Pointcoud

```python
def pc_combine(run_num,hdbscan_filepath,tripcust_filepath):   

    hdbscan_pc_file = h5py.File(hdbscan_filepath, "r") if Path(hdbscan_filepath).exists() else None
    tripcluster_pc_file = h5py.File(tripcust_filepath, "r") if Path(tripcust_filepath).exists() else None

    combined_pc_file = h5py.File(f"/Volumes/researchEXT/O16/no_efield/combined_5tracks_all/Pointcloud/run_00{run_num}.h5", "w") if run_num < 100 else h5py.File(f"/Volumes/researchEXT/O16/no_efield/combined_5tracks_all/Pointcloud/run_0{run_num}.h5", "w") if hdbscan_pc_file is not None or tripcluster_pc_file is not None else None

    if combined_pc_file is None:
        print(f"No point cloud data found for run {run_num}. Skipping.")
        return
    
    hdbscan_group = hdbscan_pc_file["cloud"] if hdbscan_pc_file else None
    tripcluster_group = tripcluster_pc_file["cloud"] if tripcluster_pc_file else None   
    
    source_group = hdbscan_group if hdbscan_group is not None else tripcluster_group
    min_event = source_group.attrs["min_event"]
    max_event = source_group.attrs["max_event"]
    
    new_group = combined_pc_file.create_group("cloud")
    new_group.attrs["min_event"] = min_event
    new_group.attrs["max_event"] = max_event

    for event in range(min_event, max_event + 1):
        event_name = f"cloud_{event}"
        if hdbscan_group is not None and event_name in hdbscan_group:
            hdbscan_data = hdbscan_group[event_name][()]
            new_data = new_group.create_dataset(event_name, data=hdbscan_data)

        elif tripcluster_group is not None and event_name in tripcluster_group:
            tripcluster_data = tripcluster_group[event_name][()]
            new_data = new_group.create_dataset(event_name, data=tripcluster_data)

        
        else:
            continue
        
        new_data.attrs["orig_run"] =  run_num #added for new spyral version
        new_data.attrs["orig_event"] = int(event) #added for new spyral version
        
        new_data.attrs["ic_amplitude"] = float(-1)
        new_data.attrs["ic_integral"] = float(-1)
        new_data.attrs["ic_centroid"] = float(-1)
        new_data.attrs["ic_multiplicity"] = float(-1)
        new_data.attrs["ic_sca_centroid"] = float(-1)
        new_data.attrs["ic_sca_multiplicity"] = float(-1)
        
    hdbscan_pc_file.close() if hdbscan_pc_file is not None else None
    tripcluster_pc_file.close() if tripcluster_pc_file is not None else None
    combined_pc_file.close()
            
            
if __name__ == "__main__":
    for run_num in range(54, 170):
        hdbscan_filepath = f"/Volumes/researchEXT/O16/no_efield/no_field_good_5_tracks/Pointcloud/run_00{run_num}.h5" if run_num < 100 else f"/Volumes/researchEXT/O16/no_efield/no_field_good_5_tracks/Pointcloud/run_0{run_num}.h5"
        tripcust_filepath =  f"/Volumes/researchEXT/O16/no_efield/tripclust_no_field_good_5_tracks/Pointcloud/run_00{run_num}.h5" if run_num < 100 else f"/Volumes/researchEXT/O16/no_efield/tripclust_no_field_good_5_tracks/Pointcloud/run_0{run_num}.h5"
        hdbscan_exists = Path(hdbscan_filepath).exists()
        tripcust_exists = Path(tripcust_filepath).exists()

        if not (hdbscan_exists or tripcust_exists):
            continue

        print(f"Processing run {run_num}...")
        pc_combine(run_num, hdbscan_filepath, tripcust_filepath)
            
```

## Combine Clustering

```python
def clus_combine(run_num,hdbscan_filepath,tripcust_filepath):   

    hdbscan_clus_file = h5py.File(hdbscan_filepath, "r") if Path(hdbscan_filepath).exists() else None
    tripcluster_clus_file = h5py.File(tripcust_filepath, "r") if Path(tripcust_filepath).exists() else None

    combined_clus_file = h5py.File(f"/Volumes/researchEXT/O16/no_efield/combined_5tracks_all/Cluster/run_00{run_num}.h5", "w") if run_num < 100 else h5py.File(f"/Volumes/researchEXT/O16/no_efield/combined_5tracks_all/Cluster/run_0{run_num}.h5", "w") if hdbscan_clus_file is not None or tripcluster_clus_file is not None else None

    if combined_clus_file is None:
        print(f"No cluster data found for run {run_num}. Skipping.")
        return
    
    hdbscan_group = hdbscan_clus_file["cluster"] if hdbscan_clus_file else None
    tripcluster_group = tripcluster_clus_file["cluster"] if tripcluster_clus_file else None   
    
    source_group = hdbscan_group if hdbscan_group is not None else tripcluster_group
    min_event = source_group.attrs["min_event"]
    max_event = source_group.attrs["max_event"]
    
    new_group = combined_clus_file.create_group("cluster")
    new_group.attrs["min_event"] = min_event
    new_group.attrs["max_event"] = max_event

    for event in range(min_event, max_event + 1):
        event_name = f"event_{event}"
        if hdbscan_group is not None and event_name in hdbscan_group:
            src = hdbscan_group[event_name]
            hdbscan_clus_file.copy(src, new_group, name=event_name)
            new_data = new_group[event_name]
            new_data.attrs["nclusters"]= src.attrs["nclusters"] 

        elif tripcluster_group is not None and event_name in tripcluster_group:
            src = tripcluster_group[event_name]
            tripcluster_clus_file.copy(src, new_group, name=event_name)
            new_data = new_group[event_name]
            new_data.attrs["nclusters"]= src.attrs["nclusters"]      
        else:
                continue
        
        new_data.attrs["orig_run"] =  run_num #added for new spyral version
        new_data.attrs["orig_event"] = int(event) #added for new spyral version
        
        new_data.attrs["ic_amplitude"] = float(-1)
        new_data.attrs["ic_integral"] = float(-1)
        new_data.attrs["ic_centroid"] = float(-1)
        new_data.attrs["ic_multiplicity"] = float(-1)
        
    hdbscan_clus_file.close() if hdbscan_clus_file is not None else None
    tripcluster_clus_file.close() if tripcluster_clus_file is not None else None
    combined_clus_file.close()
            
            
if __name__ == "__main__":
    for run_num in range(54, 170):
        hdbscan_filepath = f"/Volumes/researchEXT/O16/no_efield/no_field_good_5_tracks/Cluster/run_00{run_num}.h5" if run_num < 100 else f"/Volumes/researchEXT/O16/no_efield/no_field_good_5_tracks/Cluster/run_0{run_num}.h5"
        tripcust_filepath =  f"/Volumes/researchEXT/O16/no_efield/tripclust_no_field_good_5_tracks/Cluster/run_00{run_num}.h5" if run_num < 100 else f"/Volumes/researchEXT/O16/no_efield/tripclust_no_field_good_5_tracks/Cluster/run_0{run_num}.h5"
        hdbscan_exists = Path(hdbscan_filepath).exists()
        tripcust_exists = Path(tripcust_filepath).exists()

        if not (hdbscan_exists or tripcust_exists):
            continue

        print(f"Processing run {run_num}...")
        clus_combine(run_num, hdbscan_filepath, tripcust_filepath)
```

## Combine Estimation

```python
def est_combine(run_num, hdbscan_filepath, tripcust_filepath):
    hdbscan_df = pd.read_parquet(hdbscan_filepath) if Path(hdbscan_filepath).exists() else pd.DataFrame()
    tripcust_df = pd.read_parquet(tripcust_filepath) if Path(tripcust_filepath).exists() else pd.DataFrame()
    
    combined_df = pd.concat([hdbscan_df,tripcust_df]) if not hdbscan_df.empty and not tripcust_df.empty else hdbscan_df if not hdbscan_df.empty else tripcust_df
    combined_df_sort = combined_df.sort_values(by="event")
    
    combined_df_sort.to_parquet(f"/Volumes/researchEXT/O16/no_efield/combined_5tracks_all/Estimation/run_00{run_num}.parquet", index=False) if run_num < 100 else combined_df_sort.to_parquet(f"/Volumes/researchEXT/O16/no_efield/combined_5tracks_all/Estimation/run_0{run_num}.parquet", index=False)
    




if __name__ == "__main__":
    for run in range(54, 170):
        hdbscan_filepath = f"/Volumes/researchEXT/O16/no_efield/no_field_good_5_tracks/Estimation/run_00{run}.parquet" if run < 100 else f"/Volumes/researchEXT/O16/no_efield/no_field_good_5_tracks/Estimation/run_0{run}.parquet"
        tripcust_filepath =  f"/Volumes/researchEXT/O16/no_efield/tripclust_no_field_good_5_tracks/Estimation/run_00{run}.parquet" if run < 100 else f"/Volumes/researchEXT/O16/no_efield/tripclust_no_field_good_5_tracks/Estimation/run_0{run}.parquet"
        hdbscan_exists = Path(hdbscan_filepath).exists()
        tripcust_exists = Path(tripcust_filepath).exists()
        if not (hdbscan_exists or tripcust_exists):
            continue
        
        print(f"Processing run {run}...")
        est_combine(run, hdbscan_filepath, tripcust_filepath)
```
